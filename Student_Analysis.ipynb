## 3. Data Analysis

In this section, we analyze student academic performance
using the Student Performance dataset.

We will answer four questions:

1. What is the average final grade (G3)?
2. How many students scored above 15?
3. Is study time correlated with performance?
4. Which gender performs better on average?

### Question 1: What is the average final grade (G3)?

We calculate the average final grade of all students
using the mean() function.

G3 represents the final Mathematics grade.

```python
# Calculate the average final grade

average_grade = df["G3"].mean()

print("Average Final Grade:", average_grade)
```

### Question 2: How many students scored above 15?

We filter the students whose final grade (G3) is
greater than 15 and count the total records.

```python
# Filter students who scored above 15

students_above_15 = df[df["G3"] > 15]

# Count the number of students

count_students = len(students_above_15)

print("Students scoring above 15:", count_students)
```

### Question 3: Is study time correlated with performance?

We calculate the correlation between study time and
final grades (G3) using the corr() function.

Study time is represented by four categories:
1 = Less than 2 hours
2 = 2 to 5 hours
3 = 5 to 10 hours
4 = More than 10 hours

```python
# Calculate correlation between study time and final grade

correlation = df["studytime"].corr(df["G3"])

print("Study Time and Performance Correlation:", correlation)
```

### Question 4: Which gender performs better on average?

We group students according to gender and calculate
their average final grades using groupby() and mean().

F = Female
M = Male

```python
# Calculate average final grade for each gender

gender_average = df.groupby("sex")["G3"].mean()

print("Average Grade by Gender:")
print(gender_average)
```

```python
# Identify gender with the highest average grade

best_gender = gender_average.idxmax()

print("Gender with Higher Average Performance:", best_gender)
```

## 4. Data Visualization

In this section, we visualize student performance using
Matplotlib and Seaborn libraries.

We create three visualizations:
1. Histogram of final grades (G3).
2. Scatterplot of study time vs final grades.
3. Bar chart comparing average grades by gender.

### 4.1 Histogram of Final Grades

A histogram represents the distribution of students'
final grades (G3).

It helps us understand how many students received
grades within different score ranges.

```python
# Create a histogram of final grades

plt.figure(figsize=(8, 5))

plt.hist(df["G3"], bins=10, edgecolor="black")

plt.title("Distribution of Final Grades")
plt.xlabel("Final Grade (G3)")
plt.ylabel("Number of Students")

plt.show()
```

### 4.2 Scatterplot: Study Time vs Final Grades

A scatterplot is used to identify the relationship
between students' study time and final grades.

The X-axis represents study time.
The Y-axis represents final grades (G3).

```python
# Create scatterplot of study time vs final grades

plt.figure(figsize=(8, 5))

sns.scatterplot(data=df, x="studytime", y="G3")

plt.title("Study Time vs Final Grades")
plt.xlabel("Study Time")
plt.ylabel("Final Grade (G3)")

plt.show()
```

### 4.3 Bar Chart: Average Grade by Gender

A bar chart compares the average final grades of
male and female students.

We use groupby() to calculate the average grade
for each gender.

```python
# Calculate average grade for each gender

gender_average = df.groupby("sex")["G3"].mean()

# Create bar chart

plt.figure(figsize=(7, 5))

gender_average.plot(kind="bar", edgecolor="black")

plt.title("Average Final Grade by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Final Grade (G3)")

plt.xticks(rotation=0)

plt.show()
```
